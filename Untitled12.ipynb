Note image is an numpy array and can support np operations like slices.

```python
cropped_image = image[100:300, 50:500]  # Example of a valid crop within 531x799
```

```python
plt.imshow(cropped_image)
```

```python
image[100:200, 50:250]=(0, 0, 0)
plt.imshow(image)
```

We can add the colors to particular pixel

```python
olive=(128,128,0)
image[200:400, 50:200]=olive
plt.imshow(image)
```
